```r
df <- data.frame(
  ID = factor(1:18),
  Margarine = factor(c("B","B","B","A","B","A","B","A","B","A","B","B","A","A","A","B","A","B")),
  Before = c(6.42, 6.76, 6.56, 4.80, 8.43, 7.49, 8.05, 5.05, 5.77, 3.91, 6.77, 6.44, 6.17, 7.67, 7.34, 6.85, 5.13, 5.73),
  After4 = c(5.83, 6.20, 5.83, 4.27, 7.71, 7.12, 7.25, 4.63, 5.31, 3.70, 6.15, 5.59, 5.56, 7.11, 6.84, 6.40, 4.52, 5.13),
  After8 = c(5.75, 6.13, 5.71, 4.15, 7.67, 7.05, 7.10, 4.67, 5.33, 3.66, 5.96, 5.64, 5.51, 6.96, 6.82, 6.29, 4.45, 5.17)
)


df_long <- reshape(df,
                   varying = c("Before", "After4", "After8"),
                   v.names = "Cholesterol",
                   timevar = "Time",
                   times = c("Before", "After4", "After8"),
                   direction = "long")


model <- aov(Cholesterol ~ Margarine * Time + Error(ID/Time), data = df_long)


summary(model)
```
