```r
yield_data <- data.frame(
  Fertilizer = factor(rep(c("Blend X", "Blend Y", "Blend Z"), each = 20)),
  Crop = factor(rep(rep(c("Wheat", "Corn", "Soy", "Rice"), each = 5), 3)),
  Yield = c(
    # Blend X
    123, 156, 112, 100, 168,  # Wheat
    128, 150, 174, 116, 109,  # Corn
    166, 178, 187, 153, 195,  # Soy
    151, 125, 117, 155, 158,  # Rice
    # Blend Y
    135, 130, 176, 120, 155,
    175, 132, 120, 187, 184,
    140, 145, 159, 131, 126,
    167, 183, 142, 167, 168,
    # Blend Z
    156, 180, 147, 146, 193,
    186, 138, 178, 176, 190,
    185, 206, 188, 165, 188,
    175, 173, 154, 191, 169
  )
)

anova_model <- aov(Yield ~ Fertilizer * Crop, data = yield_data)

summary(anova_model)
```
