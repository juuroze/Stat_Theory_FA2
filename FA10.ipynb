```r
#1

X <- c(6,5,8,8,7,6,10,4,9,7)
Y <- c(8,7,7,10,5,8,10,6,8,6)

cov(X, Y)
cov(X, Y) / (sd(X) * sd(Y))

#2
#A

data(pressure)

cor(pressure$temperature, pressure$pressure)

#B

X <- pressure$temperature
Y <- pressure$pressure

X_prime <- 2*X + 6
Y_prime <- 3*Y - 15

cor(X_prime, Y_prime)


#3
#A


flights <- data.frame(
  dep_delay = c(5, -2, 10, 0, 3, 20, -5, 15, 7, 2),
  arr_delay = c(7, -1, 12, 1, 4, 25, -3, 18, 9, 3)
)

df <- na.omit(flights)

# Correlation
cor(df$dep_delay, df$arr_delay)

#B

X <- df$dep_delay
Y <- df$arr_delay

X_prime <- 2*X + 10
Y_prime <- 0.5*Y - 5

cor(X_prime, Y_prime)

#C. (EXPLANATION)

cor(X, Y)
cor(X_prime, Y_prime)

all.equal(cor(X, Y), cor(X_prime, Y_prime))

cat("The correlation will be the same because linear transformations do not change correlation.")


#D (INTERPRETATION)


r <- cor(df$dep_delay, df$arr_delay)
r

if (r > 0.7) {
  cat("Strong positive relationship: as departure delay increases, arrival delay also increases\n")
} else if (r > 0) {
  cat("Positive relationship: as departure delay increases, arrival delay tends to increase\n")
} else {
  cat("No positive relationship detected\n")
}



#4
#A


df <- data.frame(
  dep_delay = c(5, -2, 10, 0, 3, 20, -5, 15, 7, 2),
  arr_delay = c(7, -1, 12, 1, 4, 25, -3, 18, 9, 3),
  air_time = c(100, 90, 120, 110, 95, 130, 85, 125, 105, 98),
  distance = c(500, 450, 600, 550, 480, 650, 400, 620, 530, 490),
  hour = c(8, 9, 7, 6, 12, 15, 10, 18, 20, 14),
  carrier = c("AA","DL","UA","AA","WN","UA","DL","AA","WN","UA")
)


df$X1 <- ifelse(df$dep_delay <= 0, 0, 1)
df$X2 <- df$dep_delay
df$X3 <- df$air_time
df$X4 <- df$distance
df$X5 <- df$hour
df$X6 <- ifelse(df$carrier %in% c("AA","DL","UA"), 1, 0)

model <- lm(arr_delay ~ X1 + X2 + X3 + X4 + X5 + X6, data=df)

#B
summary(model)


#C

model_summary <- summary(model)

model_summary$fstatistic
pf(model_summary$fstatistic[1],
   model_summary$fstatistic[2],
   model_summary$fstatistic[3],
   lower.tail = FALSE)

# If this p-value is less than 0.05,
# reject H0 and conclude that at least one predictor is significant.


#D

model_summary <- summary(model)

f_stat <- model_summary$fstatistic

p_value <- pf(f_stat[1], f_stat[2], f_stat[3], lower.tail = FALSE)

p_value

if (p_value < 0.05) {
  cat("Reject H0: At least one predictor is significantly related to arrival delay\n")
} else {
  cat("Fail to reject H0: No significant predictors\n")
}


#E

r2 <- summary(model)$r.squared
r2


# The R^2 value represents the proportion of variability in arrival delay
# explained by the independent variables.
# A higher R^2 indicates that the model provides a better fit to the data.
```
